```python
import pandas as pd
import os
from pathlib import Path

# -------------------------------------------------
# D1.1 - Find and import the Starbucks dataset
# -------------------------------------------------

filename = "BUS303_Signature_Term_Project_Starbucks_Beverage_Menu.csv"

matches = list(Path(".").rglob(filename))

if len(matches) == 0:
    print("ERROR: Starbucks CSV file was not found.")
    print("Upload the CSV file into your Jupyter/Anaconda project, then run this cell again.")
else:
    csv_path = matches[0]
    print("File found at:", csv_path)

    sbux = pd.read_csv(csv_path)

    print("\nD1.1 - Dataset loaded successfully")
    print("Dataset shape:", sbux.shape)

    # -------------------------------------------------
    # D1.2 - Explore rows with all blank values
    # -------------------------------------------------
    print("\nD1.2 - Completely blank rows:")
    display(sbux[sbux.isnull().all(axis=1)])

    print("Number of completely blank rows:")
    print(sbux.isnull().all(axis=1).sum())

    # -------------------------------------------------
    # D1.3 - View dataframe with head()
    # -------------------------------------------------
    print("\nD1.3 - First five rows:")
    display(sbux.head())

    # -------------------------------------------------
    # D1.4 - Remove row with all blank entries
    # -------------------------------------------------
    sbux2 = sbux.dropna(how="all").copy()

    print("\nD1.4 - Blank row removed")
    print("New shape:", sbux2.shape)

    # -------------------------------------------------
    # D1.5 - Validate blank rows were removed
    # -------------------------------------------------
    print("\nD1.5 - Completely blank rows remaining:")
    print(sbux2.isnull().all(axis=1).sum())

    # -------------------------------------------------
    # D1.6 - Explore data types
    # -------------------------------------------------
    print("\nD1.6 - Data types:")
    print(sbux2.dtypes)

    print("\nDataframe info:")
    sbux2.info()

    # -------------------------------------------------
    # D1.7 - Identify incorrect data type
    # -------------------------------------------------
    print("\nD1.7 - Sugar_g current data type:")
    print(sbux2["Sugar_g"].dtype)

    print("\nProblem value in Sugar_g:")
    display(sbux2[sbux2["Sugar_g"] == "zero"])

    # -------------------------------------------------
    # D1.8 - Convert Sugar_g to float
    # -------------------------------------------------
    sbux2["Sugar_g"] = sbux2["Sugar_g"].replace("zero", "0")
    sbux2["Sugar_g"] = sbux2["Sugar_g"].astype(float)

    print("\nD1.8 - Sugar_g successfully converted")

    # -------------------------------------------------
    # D1.9 - Validate Sugar_g conversion
    # -------------------------------------------------
    print("\nD1.9 - Sugar_g new data type:")
    print(sbux2["Sugar_g"].dtype)

    # -------------------------------------------------
    # D1.10 - Explore NaN values
    # -------------------------------------------------
    print("\nD1.10 - Missing values by column:")
    print(sbux2.isnull().sum())

    print("\nRows containing NaN:")
    display(sbux2[sbux2.isnull().any(axis=1)])

    # -------------------------------------------------
    # D1.11 - Remove row containing NaN
    # -------------------------------------------------
    sbux3 = sbux2.dropna().copy()

    print("\nD1.11 - NaN row removed")
    print("New shape:", sbux3.shape)

    # -------------------------------------------------
    # D1.12 - Validate removal of NaN
    # -------------------------------------------------
    print("\nD1.12 - Total NaN values remaining:")
    print(sbux3.isnull().sum().sum())

    # -------------------------------------------------
    # D1.13 - Correct column name
    # -------------------------------------------------
    sbux3.rename(
        columns={"Caffiene_milligrams": "Caffeine_milligrams"},
        inplace=True
    )

    print("\nD1.13 - Column renamed")

    # -------------------------------------------------
    # D1.14 - Validate column name
    # -------------------------------------------------
    print("\nD1.14 - Column names:")
    print(sbux3.columns.tolist())

    # -------------------------------------------------
    # D1.15 - Write sbux3 to a new CSV file
    # -------------------------------------------------
    sbux3.to_csv(
        "Starbucks_Beverage_Menu_Cleaned.csv",
        index=False
    )

    print("\nD1.15 - Cleaned CSV saved successfully")

    # -------------------------------------------------
    # D1.16 - Reload CSV and validate
    # -------------------------------------------------
    sbux_final = pd.read_csv(
        "Starbucks_Beverage_Menu_Cleaned.csv"
    )

    print("\nD1.16 - Cleaned CSV reloaded successfully")

    display(sbux_final.head())

    print("\nCleaned dataframe shape:")
    print(sbux3.shape)

    print("\nReloaded dataframe shape:")
    print(sbux_final.shape)

    print("\nNaN values in reloaded file:")
    print(sbux_final.isnull().sum().sum())

    print("\nFinal dataframe info:")
    sbux_final.info()
```
