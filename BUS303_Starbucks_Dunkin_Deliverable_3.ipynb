```python
import pandas as pd
import matplotlib.pyplot as plt

sbux = pd.read_csv("/home/055cdef5-07dd-401f-b7ec-62aa06cfb10e/Starbucks_Beverage_Menu_Cleaned.csv")
dnkn = pd.read_csv("/home/055cdef5-07dd-401f-b7ec-62aa06cfb10e/dnkn_clean.csv")

print("Starbucks shape:", sbux.shape)
print("Dunkin shape:", dnkn.shape)
```

```python
starbucks_count = len(sbux)
dunkin_count = len(dnkn)
```

```python
print("Starbucks beverage records:", starbucks_count)
print("Dunkin beverage records:", dunkin_count)

brands = ["Starbucks", "Dunkin"]
counts = [starbucks_count, dunkin_count]

plt.bar(brands, counts)
plt.title("Beverage Portfolio Size: Starbucks vs Dunkin")
plt.ylabel("Number of Beverage Records")
plt.show()
```

The comparison shows that Starbucks has 1,203 beverage records while Dunkin has 599. Based on these cleaned datasets, Starbucks has a larger beverage portfolio. This suggests that Starbucks offers more beverage choices and variations than Dunkin, which may give customers a wider range of options.

```python
starbucks_avg_calories = sbux["Calories"].mean()
dunkin_avg_calories = dnkn["Calories"].mean()

print("Starbucks average calories:", round(starbucks_avg_calories, 2))
print("Dunkin average calories:", round(dunkin_avg_calories, 2))

brands = ["Starbucks", "Dunkin"]
avg_calories = [starbucks_avg_calories, dunkin_avg_calories]

plt.bar(brands, avg_calories)
plt.title("Average Beverage Calories: Starbucks vs Dunkin")
plt.ylabel("Average Calories")
plt.show()
```

Dunkin beverages have a higher average calorie count at about 253.33 calories, compared with about 230.83 calories for Starbucks. This suggests that, on average, Dunkin's beverages in this dataset are slightly more calorie-dense. From a competitive perspective, Starbucks may have an advantage for customers who are more focused on lower-calorie beverage options.

```python
starbucks_avg_sugar = sbux["Sugar_g"].mean()
dunkin_avg_sugar = dnkn["Sugars (g)"].mean()

print("Starbucks average sugar:", round(starbucks_avg_sugar, 2))
print("Dunkin average sugar:", round(dunkin_avg_sugar, 2))

brands = ["Starbucks", "Dunkin"]
avg_sugar = [starbucks_avg_sugar, dunkin_avg_sugar]

plt.bar(brands, avg_sugar)
plt.title("Average Sugar Content: Starbucks vs Dunkin")
plt.ylabel("Average Sugar (g)")
plt.show()
```

Dunkin beverages have a higher average sugar content at about 42.79 grams, compared with about 34.95 grams for Starbucks. This suggests that the Dunkin beverages in this dataset are generally sweeter. From a competitive perspective, Starbucks may appeal more to customers who are looking for beverages with lower average sugar content.

```python
starbucks_avg_protein = sbux["Protein_g"].mean()
dunkin_avg_protein = dnkn["Protein (g)"].mean()

print("Starbucks average protein:", round(starbucks_avg_protein, 2))
print("Dunkin average protein:", round(dunkin_avg_protein, 2))

brands = ["Starbucks", "Dunkin"]
avg_protein = [starbucks_avg_protein, dunkin_avg_protein]

plt.bar(brands, avg_protein)
plt.title("Average Protein Content: Starbucks vs Dunkin")
plt.ylabel("Average Protein (g)")
plt.show()
```

Starbucks and Dunkin both have an average protein content of about 5.99 grams per beverage. This means there is almost no difference between the two companies in terms of average protein content. Based on this dataset, protein does not appear to be a major competitive difference between Starbucks and Dunkin.

```python
starbucks_healthier = sbux[
    (sbux["Calories"] <= 100) &
    (sbux["Sugar_g"] <= 10)
]

dunkin_healthier = dnkn[
    (dnkn["Calories"] <= 100) &
    (dnkn["Sugars (g)"] <= 10)
]

print("Starbucks healthier choices:", len(starbucks_healthier))
print("Dunkin healthier choices:", len(dunkin_healthier))

brands = ["Starbucks", "Dunkin"]
healthier_counts = [len(starbucks_healthier), len(dunkin_healthier)]

plt.bar(brands, healthier_counts)
plt.title("Healthier Beverage Choices: Starbucks vs Dunkin")
plt.ylabel("Number of Beverages")
plt.show()
```

Using the criteria of 100 calories or less and 10 grams of sugar or less, Starbucks has 138 beverage choices that meet the definition of a healthier option, while Dunkin has 81. Based on these criteria, Starbucks offers more lower-calorie and lower-sugar beverage choices in this dataset, which may appeal more to health-conscious customers.

```python
plt.scatter(starbucks_avg_calories, starbucks_avg_sugar, s=150, label="Starbucks")
plt.scatter(dunkin_avg_calories, dunkin_avg_sugar, s=150, label="Dunkin")

plt.xlabel("Average Calories")
plt.ylabel("Average Sugar (g)")
plt.title("Competitive Positioning Map: Starbucks vs Dunkin")

plt.annotate(
    "Starbucks",
    (starbucks_avg_calories, starbucks_avg_sugar),
    xytext=(5, 5),
    textcoords="offset points"
)

plt.annotate(
    "Dunkin",
    (dunkin_avg_calories, dunkin_avg_sugar),
    xytext=(5, 5),
    textcoords="offset points"
)

plt.legend()
plt.show()
```

The positioning map shows that Dunkin has both a higher average calorie count and a higher average sugar content than Starbucks. Starbucks is positioned lower on both measures, which may make it more appealing to customers who are looking for beverages with fewer calories and less sugar. Dunkin, on the other hand, appears to offer beverages that are generally richer and sweeter.

```python
starbucks_zero_cal = sbux[sbux["Calories"] == 0]
dunkin_zero_cal = dnkn[dnkn["Calories"] == 0]

print("Starbucks zero-calorie beverages:", len(starbucks_zero_cal))
print("Dunkin zero-calorie beverages:", len(dunkin_zero_cal))

brands = ["Starbucks", "Dunkin"]
zero_cal_counts = [len(starbucks_zero_cal), len(dunkin_zero_cal)]

plt.bar(brands, zero_cal_counts)
plt.title("Zero-Calorie Beverage Count: Starbucks vs Dunkin")
plt.ylabel("Number of Zero-Calorie Beverages")
plt.show()
```

Starbucks has 40 zero-calorie beverage records in this dataset, while Dunkin has none. This gives Starbucks a stronger position for customers who specifically want zero-calorie drink options. Dunkin may have fewer choices for customers who are trying to avoid calories completely.

```python
starbucks_zero_items = (
    sbux[sbux["Calories"] == 0]["Product_Name"]
    .value_counts()
)

print("Starbucks zero-calorie beverage types:")
print(starbucks_zero_items)

plt.bar(starbucks_zero_items.index, starbucks_zero_items.values)
plt.title("Starbucks Zero-Calorie Beverage Options")
plt.ylabel("Number of Records")
plt.xticks(rotation=45, ha="right")
plt.show()
```

The Starbucks zero-calorie options in this dataset are mainly brewed teas. There are 10 different zero-calorie beverage types, and each appears four times, which accounts for the 40 zero-calorie records identified earlier. This shows that Starbucks offers several tea-based choices for customers who want beverages without calories.

The most caffeinated Starbucks beverage in this dataset is Brewed Coffee – True North Blend Blonde Roast at 475 mg of caffeine. The remaining top beverages are also brewed or Clover coffees, ranging from 380 to 445 mg.
```python
top5_sugar_dunkin = (
    dnkn.groupby("Item")["Sugars (g)"]
    .max()
    .sort_values(ascending=False)
    .head(5)
)

print("Top 5 most sugary Dunkin beverages:")
print(top5_sugar_dunkin)

plt.bar(top5_sugar_dunkin.index, top5_sugar_dunkin.values)
plt.title("Top 5 Most Sugary Dunkin Beverages")
plt.ylabel("Sugar (g)")
plt.xticks(rotation=45, ha="right")
plt.show()
```

The most sugary Dunkin beverages in this dataset are large frozen coffees, especially the Butter Pecan Swirl and Caramel Swirl varieties. The highest sugar content is 185 grams, which appears in two Butter Pecan Swirl Frozen Coffee options. This suggests that Dunkin’s large flavored frozen coffees contain very high amounts of sugar and may be less appealing to customers who are trying to limit sugar intake.

```python
plt.scatter(
    sbux["Calories"],
    sbux["Sugar_g"],
    alpha=0.5,
    label="Starbucks"
)

plt.scatter(
    dnkn["Calories"],
    dnkn["Sugars (g)"],
    alpha=0.5,
    label="Dunkin"
)

plt.xlabel("Calories")
plt.ylabel("Sugar (g)")
plt.title("Calories vs Sugar: Starbucks and Dunkin")
plt.legend()
plt.show()
```

The scatter plot shows a clear positive relationship between calories and sugar for both Starbucks and Dunkin beverages. In general, beverages with more calories also tend to contain more sugar. Dunkin also has several beverages with much higher calorie and sugar levels than most Starbucks beverages. This visualization shows how closely sugar content is connected to total calories and highlights some of the more extreme Dunkin beverage options.
